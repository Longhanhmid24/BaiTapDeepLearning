```python
import numpy as np

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
def sigmoid_derivative(x):
    return sigmoid(x) * (1 - sigmoid(x))

class NeuralNetwork:
    def __init__(self, x, y):
        self.x = x
        self.weights1 = np.random.rand(self.x.shape[1], 4)  # 2 input neurons, 4 hidden neurons
        self.weights2 = np.random.rand(4, 1)  # 4 hidden neurons, 1 output neuron
        self.y = y
        self.ouput = np.zeros(self.y.shape)
    def feedforward(self):
        self.layer1 = sigmoid(np.dot(self.x, self.weights1))

        self.ouput = sigmoid(np.dot(self.layer1, self.weights2))
        
    def backprop(self):
        d_weights2 = np.dot(self.layer1.T, (2 * (self.y - self.ouput) * sigmoid_derivative(self.ouput)))
        d_weights1 = np.dot(self.x.T,  (np.dot(2 * (self.y - self.ouput) * sigmoid_derivative(self.ouput), self.weights2.T) * sigmoid_derivative(self.layer1)))

        self.weights1 += d_weights1
        self.weights2 += d_weights2
        
nn = NeuralNetwork(X, y)

for i in range(10000):
    nn.feedforward()
    nn.backprop()
    
    if i % 1000 == 0:
        print (f"Epoch {i} - Loss: {np.mean(np.square(y - nn.ouput))}")
        
nn.feedforward()
print("\nKết quả dự đoán:")
print(nn.ouput)
```
