#### Ví dụ 1 với thư viện chefboost

```python
from chefboost import Chefboost as chef# Thư viện để xây dựng cây quyết định
import pandas as pd

df = pd.read_csv("/home/vophilong/Documents/2526_2A_DeepLearning/lab_part_2/lab1/lab1/dataset/golf2.txt")
df.head()
```

```python
print("Thuật toán ID3\n")

import warnings
warnings.filterwarnings("ignore")

# Thiết lập cấu hình cho thuật toán ID3
config = {'algorithm': 'ID3'}

# Huấn luyện mô hình trên tập dữ liệu huấn luyện
print("Huấn luyện mô hình trên tập dữ liệu huấn luyện:")
model = chef.fit(df[:8].copy(), config)

# Dự đoán cho một ngày cụ thể
xrow = ['Sunny',72,90,'Strong']

# Dự đoán kết quả cho ngày đã cho
prediction = chef.predict(model, xrow)
print("Dự đoán cho ngày đã cho. Có nên chơi không? {0}".format(prediction))

c = 0
print("\nSo sánh kết quả dự đoán và kết quả thực tế trên tập dữ liệu KIỂM TRA:")

# Dự đoán trên tập dữ liệu kiểm tra
for index, instance in df[8:].iterrows():
    prediction = chef.predict(model, instance)
    actual = instance['Decision']  
    print("Thực tế:", actual, "- Dự đoán:", prediction)
    
    if actual == prediction:
        c += 1

print("Độ chính xác trên tập dữ liệu kiểm tra: {0}%".format(round(c/df[8:].shape[0],4)))
```
