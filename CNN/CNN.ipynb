```python
import tensorflow as tf
import numpy as np

data = np.array([[-1.0, 0.0, 1.0]])

# sigmoid
sigmoid = tf.nn.sigmoid(data)
print("sigmoid:", sigmoid.numpy())

# tanh
tanh = tf.nn.tanh(data)
print("tanh:", tanh.numpy())

# fast sigmoid (approx)
# TensorFlow không có sẵn → tự định nghĩa
def fast_sigmoid(x):
    return x / (1 + tf.abs(x))

fast_sig = fast_sigmoid(data)
print("fast sigmoid:", fast_sig.numpy())

# softplus
softplus = tf.nn.softplus(data)
print("soft plus:", softplus.numpy())

# relu
relu = tf.nn.relu(data)
print("relu:", relu.numpy())

# softmax
softmax = tf.nn.softmax(data)
print("soft max:", softmax.numpy())
```

#### Ví dụ: Gradient (Đạo hàm)

Một hàm cần được định nghĩa bằng các biểu thức trước khi có thể tính gradient của hàm đó.

Cấu trúc grad trong Theano cho phép người dùng tạo gradient của một hàm (dưới dạng một biểu thức).

```python
import tensorflow as tf
import numpy as np

# FIX dtype
x_input = np.array([[1, 2, 3]], dtype=np.float32)
y = tf.Variable([[4, 5, 6]], dtype=tf.float32)

def f(x_np):
    x = tf.constant(x_np, dtype=tf.float32)
    z = tf.reduce_sum(((x * x) + y) * x)
    return [np.array(z.numpy())]

def g_f(x_np):
    x = tf.Variable(x_np, dtype=tf.float32)
    with tf.GradientTape() as tape:
        z = tf.reduce_sum(((x * x) + y) * x)
    grad = tape.gradient(z, x)
    return [grad.numpy()]

print("Original:", f(x_input))
print("Original Gradient:", g_f(x_input))

y.assign([[1, 1, 1]])

print("Updated:", f(x_input))
print("Updated Gradient", g_f(x_input))
```

### Neural network mô hình với 2 lớp

```python
import tensorflow as tf
import numpy as np
from sklearn.metrics import accuracy_score

# L2 regularization
def l2(x):
    return tf.reduce_sum(x ** 2)

# data
examples = 1000
features = 100
hidden = 10

D = (
    np.random.randn(examples, features).astype(np.float32),
    np.random.randint(low=0, high=2, size=examples).astype(np.float32)
)

training_steps = 1000

# variables (giống theano.shared)
w1 = tf.Variable(np.random.randn(features, hidden).astype(np.float32))
b1 = tf.Variable(np.zeros(hidden, dtype=np.float32))
w2 = tf.Variable(np.random.randn(hidden).astype(np.float32))
b2 = tf.Variable(0.0, dtype=tf.float32)

# forward
def forward(x):
    p1 = tf.nn.tanh(tf.matmul(x, w1) + b1)
    p2 = tf.nn.tanh(tf.matmul(p1, tf.reshape(w2, (-1, 1))) + b2)
    return tf.squeeze(p2)

# predict
def predict(x):
    p2 = forward(x)
    return p2 > 0.5

# loss
def compute_loss(x, y):
    p2 = forward(x)
    error = tf.keras.losses.binary_crossentropy(y, p2)
    loss = tf.reduce_mean(error) + 0.01 * (l2(w1) + l2(w2))
    return loss, p2, error

# train step
def train(x, y):
    with tf.GradientTape() as tape:
        loss, p2, error = compute_loss(x, y)
    grads = tape.gradient(loss, [w1, b1, w2, b2])

    # update (giống Theano updates)
    w1.assign_sub(0.1 * grads[0])
    b1.assign_sub(0.1 * grads[1])
    w2.assign_sub(0.1 * grads[2])
    b2.assign_sub(0.1 * grads[3])

    return p2, error

# --- TRAINING ---

# trước training
pred_before = predict(D[0]).numpy()
print("Accuracy before Training:", accuracy_score(D[1], pred_before))

# train loop
for i in range(training_steps):
    p2, error = train(D[0], D[1])

# sau training
pred_after = predict(D[0]).numpy()
print("Accuracy after Training:", accuracy_score(D[1], pred_after))
```
